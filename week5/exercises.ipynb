```python
import numpy as np
import scipy as sp
import matplotlib.pyplot as plt
```

```python
# length of a curve
N = 1000
t = np.linspace(0, 1, N+1)
c = lambda t: np.array([2*t+1, -t**2])

length = np.sum([np.linalg.norm(c(t[i]) - c(t[i-1])) for i in range(1, N)])
print(f"Approximation using equation 4.2: {length:.3f}")

print(f"Analytically: {(np.sqrt(2)+np.log(1+np.sqrt(2))):.3f}")
```

```python
# using eq.4.5
# length of a curve
N = 100000
t = np.linspace(0, 1, N+1)
c = lambda t: np.array([2*t+1, -t**2])
dct = lambda t: np.array([2, -2*t])

speeds = np.sum([np.linalg.norm(dct(t[i])) for i in range(1, N)])
length = np.sum(speeds) * (1/N) # we discretize the interval [0, 1] into N steps, so the length is the sum of the speeds at each step divided by N
print(f"Approximation using equation 4.5: {length:.3f}")
```

```python
def curve_length(c, t0=0.0, t1=1.0, N=10000):
    t_values = np.linspace(t0, t1, N+1)
    points = np.array([c(t) for t in t_values])
    diffs = np.diff(points, axis=0)
    return np.sum(np.linalg.norm(diffs, axis=1))

# Example: Second-order polynomial curve in latent space
def latent_curve(t):
    # Example: c(t) = [a*t^2 + b*t + c, d*t^2 + e*t + f]
    a, b, c0, d, e, f0 = 1.0, 0.5, 0.0, -0.5, 1.0, 0.5
    return np.array([a*t**2 + b*t + c0, d*t**2 + e*t + f0])

length_latent = curve_length(latent_curve)
print("Length of the latent curve:", length_latent)
```
